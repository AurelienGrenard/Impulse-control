# Harvesting — unlimited impulses ($d=6$)

Load a trained impulse-control model and inspect its numerical results.


## Train your own model

Run the following command from the project root:

```bash
python -m impulse_control.train_harvesting --mode unlimited --dimension <DIMENSION> --output runs/harvesting_unlimited_d<DIMENSION>.pt
```

The training script saves the checkpoint specified by `--output`. The supplied pretrained models can be loaded directly without retraining.

````python
from impulse_control.saving import load_all_results_unlimited

all_results_loaded = load_all_results_unlimited("runs/harvesting_unlimited_d6.pt", map_location="cpu")
````

## Value functions

Compare finite-horizon neural values with the closed-form infinite-horizon solution.

````python
from impulse_control.plotting import plot_unlimited_summary

_ = plot_unlimited_summary(all_results_loaded, output="figures/harvesting_unlimited_d6_value_functions.png")
````

## Policy consistency

Compare Monte Carlo scores under the learned and exact band policies across horizons. Error bars show standard errors.

````python
from impulse_control.plotting import plot_unlimited_policy_consistency

_ = plot_unlimited_policy_consistency(all_results_loaded, output="figures/harvesting_unlimited_d6_policy_consistency.png")
````
