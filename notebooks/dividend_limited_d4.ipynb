# Dividend — limited impulses ($d=4$)

Load a trained impulse-control model and inspect its numerical results.


## Train your own model

Run the following command from the project root:

```bash
python -m impulse_control.train_dividend --mode limited --dimension <DIMENSION> --seed 1234 --output runs/dividend_limited_d<DIMENSION>.pt
```

The training script saves the checkpoint specified by `--output`. The supplied pretrained models can be loaded directly without retraining.

````python
from impulse_control.saving import load_results_bundle

bundle = load_results_bundle("runs/dividend_limited_d4.pt", map_location="cpu")
all_results_loaded = bundle["bounded_results"]
````

## Value functions

Compare each finite impulse budget with the unconstrained value function.

````python
from impulse_control.plotting import plot_limited_summary

_ = plot_limited_summary(bundle, output="figures/dividend_limited_d4_value_functions.png")
````

## Controlled paths

Compare impulse budgets on the same Brownian realization. Vertical jumps mark interventions.

````python
from impulse_control.plotting import plot_limited_paths

_ = plot_limited_paths(bundle, seed=123, output_prefix="figures/dividend_limited_d4")
````
